# Recorrido por `numerica`

Un paseo corto por raíces, cuadratura y una ecuación diferencial. Los mismos cálculos están en los scripts de `examples/`.

```python
import math

from numerica import bisection, newton, rk45, romberg

raiz = bisection(lambda x: x**3 - x - 1, 1.0, 2.0)
print(raiz)
print(newton(lambda x: x**3 - x - 1, 1.5, df=lambda x: 3 * x**2 - 1))
```

```python
integral = romberg(math.sin, 0.0, math.pi, tol=1e-12)
print(integral, "error", abs(integral.value - 2.0))
```

```python
solucion = rk45(lambda t, y: -y, (0.0, 2.0), 1.0, rtol=1e-7)
print(solucion)
print("y(2) ≈", solucion.y[-1], "exacto", math.exp(-2.0))
```
